## Load h5ad

```python
base_path='/content/drive/MyDrive/Colab Notebooks/Haley/YSL/'
```

```python
adata = sc.read(base_path+'YSL-full_pp_norm.h5ad')
```

## Annotate by ZMAP_Tissue

```python
adata.layers.keys()
```

```python
# Automated annotation with: ZMAP_Tissue
adata = zmap.predict.annotate_with_zmap(
    adata,
    query_raw_counts_source='raw_nolog',
    ref_label_col='ZMAP_Tissue',
    query_label_col='leiden',
    predict_kwargs = {'p_thresh': 0.5},
    verbosity=3
)
```

## Annotate by ZMAP_CellType

```python
# Automated annotation with: ZMAP_CellType
adata = zmap.predict.annotate_with_zmap(
    adata,
    query_raw_counts_source='raw_nolog',
    ref_label_col='ZMAP_CellType',
    query_label_col='leiden',
    predict_kwargs = {'p_thresh': 0.5},
    verbosity=3
)
```

## Save ZMAP annotated h5ad

```python
adata.write(base_path+'260615_1_YSL_cluster_analyzed_zmap.h5ad')
```
